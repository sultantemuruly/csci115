## (15 points) Task 1 - load accounts

Here we have the dataset file (accounts.csv) which stores the sets of ID and password

* load_accounts(filename)  # this function load the data from the file. 
Load a data file and acquire the ID_DB in this form Dictionary form ID:PASSWORD. 
```python
ID_DB has {'admin':'123456', 'Robert':'qwer123', 'Michael':'rtyu123', 'Thomas':'q1w2e3'}
```
Each ID/PASS must be acquired from the FILE. 

````python
import csv

def load_accounts(filename):
    id_db = {}
    with open(filename, 'r') as f:
        reader = csv.reader(f)
        rows = list(reader)
        rows.pop(0)
        for row in rows:
            id_db[row[0]] = row[1]
    return id_db

ID_DB = load_accounts('accounts.csv')
print(ID_DB)
````

## (15 points) Task 2 - log_in()

We have global variables and these are initialized as follows:
```python

logged_in_id = ''
```

<font color='blue'>log_in()</font> function will follow the logic below:

Step 1 : Ask the user to enter the ID.
* If the ID is existing in the DB, then go to Step 2.
* If the ID is not existing in the DB, print out <font color='green'>"Your ID cannot be found"</font> and <font color='blue'>terminate a function</font>.

Step 2 : Ask the user to enter password.
* If the user correctly enter the password with respect to the ID, then print out <font color='green'>"you are logged-in  with ID: {ID}"</font>. The global variables <font color='green'>'log_in_state'</font> and <font color='green'>'logged_in_id'</font> should be also updated into <font color='green'>'True'</font> and <font color='green'>current ID</font>, respectively.
```python
log_in_state = True
logged_in_id = YourID
```
* If the user entered a wrong password, then <font color='blue'>terminate a function</font>.

````python
log_in_state = False
logged_in_id = ''

def log_in():
    input_id = input("Enter your ID: ")
    
    if input_id not in ID_DB:
        print("Your ID cannot be found")
        return

    password = input("Enter your password: ")
    
    if password != ID_DB[input_id]:
        print("Password is wrong")
        return

    print(f"you are logged-in with ID: {input_id}")
    global log_in_state, logged_in_id
    log_in_state = True
    logged_in_id = input_id
````
