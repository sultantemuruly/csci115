```python
def check_is_won(moves_list):
    winning_combinations = [
        ["A1", "A2", "A3"],
        ["B1", "B2", "B3"],
        ["C1", "C2", "C3"],
        ["A1", "B1", "C1"],
        ["A2", "B2", "C2"],
        ["A3", "B3", "C3"],
        ["A1", "B2", "C3"],
        ["A3", "B2", "C1"]
    ]

    for combination in winning_combinations:
        count = 0
        for cell in combination:
            if cell in moves_list:
                count += 1
        if count == 3:
            return True

    return False
```

```python
test_list1 = ['A2', "A1", 'A3']
test_list2 = ['B2', "C3", 'A1']
test_list3 = ['B3', "C3", 'A3']
test_list4 = ['B3', "C3", 'A2']
test_list5 = ['B3', "C3"]
test_list6 = ["C3"]

print(check_is_won(test_list1))
print(check_is_won(test_list2))
print(check_is_won(test_list3))
print(check_is_won(test_list4))
print(check_is_won(test_list5))
print(check_is_won(test_list6))
```

```python
def tictactoe_result(moves: list[str]) -> str:
    i = 0
    o_moves = []
    x_moves = []

    while i < len(moves):
        current_move = moves[i]

        if i % 2 == 0:
            # O's turn
            if len(o_moves) == 4:
                o_moves.pop(0)   # remove oldest O mark first
            o_moves.append(current_move)

            print(f"O moves: {o_moves}")
            if check_is_won(o_moves):
                print(f"O moves: {o_moves}")
                return "O wins"

        else:
            # X's turn
            if len(x_moves) == 4:
                x_moves.pop(0)   # remove oldest X mark first
            x_moves.append(current_move)

            print(f"X moves: {x_moves}")
            if check_is_won(x_moves):
                print(f"X moves: {x_moves}")
                return "X wins"

        i += 1

    # No winner after all given moves
    if len(moves) >= 30:
        return "Draw"
    else:
        return "Game not finished"
```

```python
moves = ["A1", "B2", "A2", "A3", "C1", "B1", "B3", "C2", "C3"]

tictactoe_result(moves)
```
