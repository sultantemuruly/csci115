## Problem 4: Using `tell()` and `seek()` [15%]
Write a Python program that does the following:
- Opens a `user_info.txt` file in read mode.
- Reads and prints the first 10 characters.
- Uses `tell()` to display the current position in the file.
- Moves the file pointer (file object) back to the beginning using `seek()`.
- Reads and prints the first 5 characters again to confirm that the pointer was reset.

### Requirements
You have to
- use `with` context manager
- use `tell()` and `seek()`
- use `read()` function

### Example input file contents
```
Name: John Doe, Age: 28
Hobbies: Reading, Swimming, Traveling
```
### Expected output
```terminal
First 10 character in file: Name: John
Current position in file: 11
First 5 characters in file: Name:

````python
with open('user_info.txt', 'r') as f:
    print('First 10 character in file:', f.read(10))
    print('Current position in file:', f.tell())
    f.seek(0)
    print('First 5 character in file:', f.read(5))
````

## Problem 5: Count character frequency in a file [30%]
Write a Python program that:
- Opens `user_info.txt` in reading mode.
- Reads the content of the file.
- Counts the frequency of each character, ignoring spaces and newlines.
- Prints the character frequency.

### Requirements
You have to
- use `for` loop to iterate over text
- use `dict` to store and update frequency of each character, where characters are `key`s and corresponding frequencies are `value`s
- use `print()`'s sep to print each character and its frequency

### Example input file contents
```
Name: John Doe, Age: 28
Hobbies: Reading, Swimming, Traveling
```
### Expected output
```terminal
frequency of characters:
N -> 1
a -> 3
m -> 3
e -> 6
: -> 3
J -> 1
o -> 3
h -> 1
n -> 4
D -> 1
, -> 3
A -> 1
g -> 4
2 -> 1
8 -> 1
H -> 1
b -> 2
i -> 5
s -> 1
R -> 1
d -> 1
S -> 1
w -> 1
T -> 1
r -> 1
v -> 1
l -> 1


````python
with open('user_info.txt', 'r') as f:
    letters = {}
    for line in f:
        line = line.strip().replace(' ', '')
        
        for s in line:
            if s in letters:
                letters[s] += 1
            else:
                letters[s] = 1

print('Frequency of characters:')
for letter, count in letters.items():
    print(letter, count, sep=' -> ')
````
