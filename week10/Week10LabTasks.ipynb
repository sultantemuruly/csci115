## Problem 2: Student Gradebook [File Writing]

You will build a small gradebook system that collects student data from the user, writes it to a file, and then reads it back.

### Task 2.1: Collect and write student data

1. Ask the user how many students to enter (use `input()`).
2. For each student, ask for:
   - Student name
   - Three exam scores (as integers)
3. For each student, compute their **average score**.
4. Write all data to a file called `gradebook.txt` using the `print()` function with the `file` parameter.

**File format (each student on one line):**
```
Name: Alice | Scores: 85, 90, 78 | Average: 84.3
Name: Bob | Scores: 72, 68, 80 | Average: 73.3
```

**Requirements:**
- Use `print()` with `sep` and `file` parameters to write to the file
- Don't use f-strings or `.format()`
- Don't use `write()` for this task
- Close the file properly after writing

### Example interaction:
```
How many students? 2
Enter name for student 1: Alice
Enter score 1 for Alice: 85
Enter score 2 for Alice: 90
Enter score 3 for Alice: 78
Enter name for student 2: Bob
Enter score 1 for Bob: 72
Enter score 2 for Bob: 68
Enter score 3 for Bob: 80
```

````python
students_num = int(input("How many students? "))
f = open('gradebook.txt', 'w')

for i in range(students_num):
    name = input("Enter name for student " + str(i+1) + ": ")
    score_1 = int(input('Enter score 1 for ' + name + ': '))
    score_2 = int(input('Enter score 2 for ' + name + ': '))
    score_3 = int(input('Enter score 3 for ' + name + ': '))
    avg = (score_1 + score_2 + score_3) / 3
    print('Name: ' + name, "Scores: " + str(score_1), str(score_2), str(score_3), "Average: " + str(avg), sep=' | ', file=f)
f.close()
````

### Task 2.2: Read and display the gradebook

Open `gradebook.txt` in read mode and display its contents.

1. Use `readlines()` to read all lines.
2. Print each line **without extra blank lines** (strip the trailing newline).
3. After printing all students, print the total number of students in the file.

**Expected output (based on example above):**  

```
--- Gradebook Contents ---
Name: Alice | Scores: 85, 90, 78 | Average: 84.3
Name: Bob | Scores: 72, 68, 80 | Average: 73.3
--------------------------
Total students: 2
```

**Requirements:**
- Use `with` context manager to open the file
- Use `readlines()`
- Use `.strip()` to remove trailing newlines

````python
f = open('gradebook.txt', 'r')
lines = f.readlines()

print('', 'Gradebook Contents', '', sep='---')
for line in lines:
    print(line, end='')
print('-'*30)
print('Total students: ' + str(len(lines)))

f.close()
````
