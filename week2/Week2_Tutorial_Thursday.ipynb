Sources:

* Python slice string syntax is:

    - str_object[start_index:end_index:step_size]
* All these parameters are optional - start_pos default value is 0, the end_pos default value is the length of string and step default value is 1.






````python
s = 'HelloWorld'

s1 = s[::3]
print(s1)

s1 = s[5:]
print(s1)

s1 = s[8:1:-1]
print(s1)

# end to start

s1 = s[8:1:-2]
print(s1)

s1 = s[-10]
print(s1)

a = "hello eve\rryone"
print(a)
````

### Handling Escape Sequences in Strings

**R Means 'Raw String'**

* Let's simply add 'r' to the string.

* Normally, Python uses backslashes as escape characters. Prefacing the string definition with 'r' is a useful way to define a string where you need the backslash to be an actual backslash and not part of an escape code that means something else in the string.

- **Example**:
    ```python
    print(r"C:\Users\Name\nNewFolder")
    ```
    **Output**:
    ```
    C:\Users\Name\nNewFolder
    ```

### Double Backslashes
Alternatively, you can use double backslashes to represent a single backslash in a normal string.

- **Example**:
    ```python
    print("C:\\Users\\Name")
    ```
    **Output**:
    ```
    C:\Users\Name
    ```
