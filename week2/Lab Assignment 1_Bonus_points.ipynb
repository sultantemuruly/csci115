## 2) Indexing (getting a single character)
Each character has a position called an **index**.

- First character index is `0`
- You can use negative indices from the end: `-1` is the last character

Example:
`text[0]` → first character
`text[-1]` → last character


````python
text = "abracadabra"
print(text[0])     # first character
print(text[3])     # 4th character
print(text[-1])    # last character
print(text[-2])    # second last
````

### Common mistake
Index out of range happens when you access an index that doesn't exist.
Try it (and then undo):
`text[100]`


## 3) Slicing (getting a substring)
Slicing uses the pattern:

`text[start : stop : step]`

- `start` is **included**
- `stop` is **not included**
- `step` is how many characters to jump each time

Useful shortcuts:
- `text[:stop]` → from start
- `text[start:]` → to the end
- `text[:]` → copy
- `text[::2]` → every 2nd character
- `text[::-1]` → reversed string


````python
text = "abracadabra"
print(text[0:4])   # 'abra'
print(text[4:7])   # 'cad'
print(text[:4])    # from start
print(text[7:])    # to end
print(text[::2])   # step
print(text[::-1])  # reverse
````

## 4) Useful string tools
- `len(text)` gives length
- `'a' in text` checks if substring exists
- `.lower()`, `.upper()` change case
- `.strip()` removes spaces at ends
- `.replace(old, new)` replaces parts


````python
text = "  Hello World  "
print(len(text))
print("World" in text)
print(text.lower())
print(text.upper())
print(text.strip())
print(text.replace("World", "Python"))
````

## 5) String formatting
String formatting means building a string with values inside it.

### A) f-strings (recommended)
Use an `f` before the quotes:
`f"Hello {name}"`

### B) `str.format()`
`"Hello {}".format(name)`

### C) Old `%` formatting (you may see it in old code)
`"Hello %s" % name`


````python
name = "Ada"
age = 17

print(f"My name is {name} and I am {age} years old.")
print("My name is {} and I am {} years old.".format(name, age))
print("My name is %s and I am %d years old." % (name, age))
````

## 6) Formatting numbers
You can control decimals, width, alignment, and more.

Examples with f-strings:
- `{value:.2f}` → 2 decimals
- `{value:>10}` → right-align in width 10
- `{value:<10}` → left-align in width 10
- `{value:^10}` → center in width 10
- `{value:,}` → thousands separators
- `{value:.0%}` → percentage (0 decimals)


````python
price = 1234.56789
ratio = 0.2375

print(f"Price (2 decimals): {price:.2f}")
print(f"With commas: {price:,.2f}")
print(f"As percent: {ratio:.0%}")

word = "cat"
print(f"Right: '{word:>10}'")
print(f"Left : '{word:<10}'")
print(f"Center: '{word:^10}'")
````

## Task 6 — Formatting an invoice line
You have a product name, quantity, and unit price.

Print a single line like this (alignment matters):
- Product name left-aligned width 15
- Quantity right-aligned width 5
- Unit price right-aligned width 10 with 2 decimals
- Total price right-aligned width 10 with 2 decimals

Example output:
```
Apples              3      2.50      7.50
```


````python
product = "Apples"
qty = 3
unit_price = 2.5

total_price = qty * unit_price

out = f"{product:<15} {qty:>5} {unit_price:>10.2f} {total_price:>10.2f}"
print(out)
````

## Task 7 — Find and slice around a word
Given a sentence, find the position of the first occurrence of the word `'Python'`.
Then print:
1) Everything **before** `'Python'`
2) The word `'Python'` itself
3) Everything **after** `'Python'`

Hint: use `.find()` and slicing.


````python
sentence = "I started learning Python last year. Python is great!"

first_index = sentence.find("Python")
final_index = first_index + len("Python")

print(sentence[:first_index])
print(sentence[first_index:final_index])
print(sentence[final_index:])
````

## Task 8 — Temperature report (f-strings)
Given a city and temperature in Celsius:
1) Convert to Fahrenheit: `F = C * 9/5 + 32`
2) Print a report using f-strings with **1 decimal** for both values.

Example:
`"Pavlodar",  -12.3` → `"Pavlodar: -12.3°C = 9.9°F"`


````python
city = "Pavlodar"
c = -12.3


F = (c * 9/5) + 32
print(f"{city}: {F:.2f}°F.")
````
