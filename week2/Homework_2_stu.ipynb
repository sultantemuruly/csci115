## Part I – string input/output and mathematic operators

### Question 1
Ask the user for the total number of papers submitted in a class. Ask the user for the number of students in the class. Output the average number of papers submitted per student. Assume class has more than zero students, and all numbers are integers.

Expected output: 


````python
papers = int(input("How many papers were submitted: "))
students = int(input("How many students are in class: "))
average = papers / students

print(f"The average number of papers submitted per student is: {average}")
````

### Question 2
Prompt the user for three numbers and then output their sum, difference, product, quotient, remainder, average. First number always before second number and second before third for each operation. Assume the second and third numbers will not be zero.

Expected output: 


````python
first_num = float(input("Enter the first number: "))
second_num = float(input("Enter the second number: "))
third_num = float(input("Enter the third number: "))

sum_nums = first_num + second_num + third_num
difference = first_num - second_num - third_num
product = first_num * second_num * third_num
quotient = first_num / second_num / third_num
remainder = first_num % second_num % third_num
average = sum_nums / 3

print(f"The sum of {first_num}, {second_num}, and {third_num} is: {sum_nums}")
print(f"The difference of {first_num}, {second_num}, and {third_num} is: {difference}")
print(f"The product of {first_num}, {second_num}, and {third_num} is: {product}")
print(f"The quotient of {first_num}, {second_num}, and {third_num} is: {quotient}")
print(f"The remainder of {first_num}, {second_num}, and {third_num} is: {remainder}")
print(f"The average of {first_num}, {second_num}, and {third_num} is: {average}")
````

### Question 3
Ask the user for 3 words/phrases. Output all possible combinations of the 3 strings. (first + second + third, third + second + first, etc.)

````python
first = input("Enter the first word: ")
second = input("Enter the second word: ")
third = input("Enter the third word: ")

print(first + " " + second + " " + third)
print(first + " " + third + " " + second)
print(second + " " + first + " " + third)
print(second + " " + third + " " + first)
print(third + " " + first + " " + second)
print(third + " " + second + " " + first)
````

### Question 4
Given a string of odd length greater than 7, return a string made of the middle three chars of a given String. Your code should work for any odd length String.

Expected output: 


````python
word = input("Original String is ")

first_index = len(word)//2 - 1
last_index = len(word)//2 + 2

print("Middle three characters are: " + word[first_index:last_index])
````

### Question 5
Given 2 strings, s1 and s2, create a new string by appending s2 in the middle of s1

Expected output: 


````python
s1 = input("")
s2 = input("")

middle = len(s1) // 2
out = s1[:middle] + s2 + s1[middle:]
print(out)
````

### Question 6
Given 2 strings, s1, and s2 return a new string made of the first, middle and last char each input string

Expected output: 


````python
s1 = input("")
s2 = input("")

first_s1 = s1[0]
middle_s1 = s1[len(s1)//2]
last_s1 = s1[-1]

first_s2 = s2[0]
middle_s2 = s2[len(s2)//2]
last_s2 = s2[-1]

print(first_s1 + first_s2 + middle_s1 + middle_s2 + last_s1 + last_s2)
````

## Part III - String formatting

### Question 7. 

From the given variables, output the sentence using each of the String Formatting methods.
1) placeholders {}
2) f-strings
3) % operator
``` python
city = "New York"
country = "USA"
population = 8175133
```

Expected output:




````python
city = "New York"
country = "USA"
population = 8175133
# basic
print("The city of",city, "is located in the",country, "and has a population of",population)
# placeholder with format function
print("The city of {} is located in the {} and has a population of {}".format(city, country, population))
# f-strings
print(f"The city of {city} is located in the {country} and has a population of {population}")
# % operator
print("The city of %s is located in the %s and has a population of %d" % (city, country, population))
````
