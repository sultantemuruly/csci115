# Task 1: Sum of First N Numbers (Direct Recursion)

## Problem Statement
Write a recursive function `sum_to_n(n)` that returns:

\[
1 + 2 + 3 + ... + n
\]

### Requirements
- Use recursion (no loops)
- If `n` is not an integer or `n < 0`, return `"Invalid input"`

## Expected Output
```text
sum_to_n(5) -> 15
sum_to_n(1) -> 1
sum_to_n(0) -> 0
sum_to_n(-3) -> Invalid input

````python
def sum_to_n(n):
    if n < 0:
        print('Invalid input')
        return None
    if n == 0:
        return 0
    return n + sum_to_n(n - 1)
````

````python
sum_to_n(10)
````

# Task 2: Divisible by 3 (Indirect Recursion)

## Problem Statement
Implement **indirect recursion** using two functions:
- `is_multiple_of_3(n)`
- `reduce_by_3(n)`

Rules:
- `is_multiple_of_3(n)` returns `True` if `n` is divisible by 3, else `False`
- Use recursion by repeatedly reducing the number by 3
- The two functions must call each other (mutual recursion)

### Requirements
- Accept only integers `n >= 0`
- If invalid input, return `"Invalid input"`

## Expected Output
```text
is_multiple_of_3(9) -> True
is_multiple_of_3(10) -> False
is_multiple_of_3(0) -> True
is_multiple_of_3(-1) -> Invalid input

````python
def is_multiple_of_3(n):
    if n < 0:
        print('Invalid input')
        return None
    if n == 0:
        return True
    elif n < 3:
        return False
    return reduce_by_3(n)

def reduce_by_3(n):
    next_number = n - 3
    return is_multiple_of_3(next_number)
````

````python
is_multiple_of_3(9)
````

# Task 3: Sum of Digits (Non-Tail Recursion)

## Problem Statement
Write a recursive function that returns the **sum of digits** of a non-negative integer.

Example:
- `sum_digits(725) = 7 + 2 + 5 = 14`

## Requirement (Important)
You must implement this using **non-tail recursion**.

## Requirements
- Accept only integers `n >= 0`
- If invalid input, return `"Invalid input"`
- Use recursion (no loops)

## Concept Reminder
- **Non-tail recursion:** does extra work after the recursive call returns  
  Pattern: `return x * recursive_call(...)`
- **Tail-recursion style:** recursive call is the last thing executed  
  Pattern: `return recursive_call(updated_values)`

## Expected Output
```text
sum_digits(725) -> 14
sum_digits(0) -> 0
sum_digits(9) -> 9
sum_digits(-3) -> Invalid input


````python
def sum_digits(n):
    if n < 0:
        print('Invalid input')
        return None
    if n > 10:
        last_digit = n % 10
        remaining_digits = n // 10
        return last_digit + sum_digits(remaining_digits)
    else:
        return n
````

````python
sum_digits(725)
````
