## Problem 1. Working with `pickle`

### Task 1.1: Save student records

Create a list of **3 student records**, where each record is a list in the form:

`[student_id, name, grade]`

Example:
`[101, "Emma", 89.5]`

Using the `pickle` module, save **each record separately** into a file called `students.dat`.

#### Requirements:
- import `pickle`
- open the file in binary write mode
- use `pickle.dump()`
- use `with`

```python
import pickle

students = [[0, 'John', 5], [1, 'Joseph', 4], [2, 'Stacy', 5]]

with open('students.dat', 'wb') as file:
    for student in students:
        pickle.dump(student, file)
```

### Task 1.2: Load records

Read all records back from `students.dat` and print them one by one.

#### Requirements:
- open the file in binary read mode
- use `pickle.load()`
- use a loop
- stop reading when the file ends

```python
with open('students.dat', 'rb') as file:
    while True:
        try:
            student = pickle.load(file)
            print(student)
        except EOFError:
            break
```

## Problem 2. Handling Exceptions

### Task 2.1: Input validation

Write a function called `get_integer()` that keeps asking the user to enter an integer until the input is valid.

Then call the function and print the accepted number.

#### Requirements:
- use `try` and `except`
- handle `ValueError`

```python
def get_integer():
    while True:
        try:
            num = int(input('Enter an integer: '))
            return num
        except ValueError:
            print('Error')
```

### Task 2.2: Safe division

Write a function called `safe_division()` that:

1. asks the user for two integers  
2. divides the first number by the second  
3. handles possible errors

#### Requirements:
- handle `ValueError`
- handle `ZeroDivisionError`
- use `else`
- use `finally` to print: `Operation finished.`

```python
def safe_division():
    try:
        num1 = get_integer()
        num2 = get_integer()
        result = num1 / num2
        print(f'{num1} / {num2} = {result}')
    except ValueError:
        print('Incorrect value')
    except ZeroDivisionError:
        print('Division by zero')
    finally:
        print('Operation finished')
```

### Task 2.3: Access by index

Write a function called `show_student_by_index(students)`.

The function should:
1. ask the user for an index
2. print the student record at that index

Use the list of students loaded in Problem 1.

#### Requirements:
- handle `ValueError` if the index is not an integer
- handle `IndexError` if the index is out of range

```python
def show_student_by_index(students):
    try:
        index = int(input(""))
        student = students[index]
        print(student)
    except ValueError:
        print('Incorrect value')
    except IndexError:
        print('Index out of bounds')
```

## Problem 3. Expense Logger

### Task 3.1: Save expenses and show the total

Write a program that repeatedly asks the user to enter:

- an expense category,
- an expense amount.

The user should be able to stop the input by entering `"stop"` as the category.

Each valid expense should be saved into a file called `expenses.dat` using the `pickle` module.
Each record should be stored in the form:

`[category, amount]`

After all input is finished, load the records from the file and print:

- all saved expenses,
- the total amount spent.

#### Requirements:
- use `pickle.dump()` to save records,
- use `pickle.load()` to read records,
- handle `ValueError` if the amount is not a valid number,
- use a loop and stop reading with `EOFError`.

```python
import os
import pickle

while True:
    try:
        category = input()
        if category == 'stop':
            break
        amount = int(input())
        
        if not os.path.isfile("expenses.dat"):
            with open('expenses.dat', 'wb') as file:
                pickle.dump((category, amount), file)
        else:
            with open('expenses.dat', 'ab') as file:
                pickle.dump((category, amount), file)
    except ValueError:
        print('Error')

with open('expenses.dat', 'rb') as file:
    while True:
        try:
            category, amount = pickle.load(file)
            print(f'Category: {category}, Amount: {amount}')
        except EOFError:
            break
```
