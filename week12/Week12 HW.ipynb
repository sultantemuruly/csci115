1. Write a program that reads a file and prints the first line of the file. If the file does not
exist, catch the exception and print a custom error message.

```python
try:
    with open('file.txt', 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found.")
```

2. Write a program which defines a dictionary of 5 items (key-value pairs), asks the user to
enter a key, and prints the value of that key. Your program should not throw an exception when the user
enters a non-existing key, rather it should ask the user to try another key. Use the `try` and `except` statements to catch the `KeyError` exception<sup>1</sup>. Note that if you use the `get()` method of a dictionary to access an item and if the specified key does not exist, it will not raise an exception but will return `None`. Therefore, in this task, use `d["key"]` to access the item.

```python
my_dict = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}

while True:
    try:
        key = int(input("Enter a key: "))
        print(my_dict[key])
        break
    except KeyError:
        print("Key not found. Try another key.")
```

3. Modify the following program to handle different exceptions such as `FileNotFoundError`
(may raise in line 15), `ValueError` (lines 4-6), and `EOFError` (line 17) <sup>1</sup>.

```python
import pickle

file = open("students.pkl","ab")
try:    
    while True:
        sid = int(input("Student ID: "))
        name = input("Student Name : ")
        grade = float(input("Grade: "))
        record=[sid, name, grade]
        pickle.dump(record, file)
        stop = input("Enter more records (y/n)? ")
        if stop.lower() == "n":
            break
    print("Size of file is {} bytes:".format(file.tell()))
    file.close()
except ValueError:
    print("Incorrect Value.")

try:
    file = open("students.pkl","rb")
    while True:
        data = pickle.load(file)
        print(data)
    file.close()
except FileNotFoundError:
    print("File not found.")
except EOFError:
    print("End of file reached.")
```

1. Recall that _KeyError_ occurs when a key does not exist in a dictionary, _FileNotFoundError_
occurs when a program tries to read a file which does not exist, _ValueError_ occurs when
there is a wrong value in a specified data type, and _EOFError_ occurs when the file position
reaches the end of file while reading a file. For more exception types, see
https://www.w3schools.com/python/python_ref_exceptions
