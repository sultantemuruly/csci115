### Task 1 

Given a list consisting of integer values nums. Find all the elements of the list that have positive values and are divided to either 3, 5 or 7. All the found elements should be copied into another list in the reverse order (first goes elements that satisfy conditions and located closer to the end of the list nums). 

**Example:** 
- Input: nums = [10, 15, -12, 24, 17, -6, 6, 18] 
- Output : found = [18, 6, 24, 15,10]

````python
nums = [10, 15, -12, 24, 17, -6, 6, 18]
found = []

for num in nums:
    if num < 0:
        continue
    if num % 3 == 0:
        found.insert(0, num)
    elif num % 5 == 0:
        found.insert(0, num)
    elif num % 7 == 0:
        found.insert(0, num)

print(found)
````

### Task 2 

For the given string password, check the following conditions:
- a) The string contains at least 2 symbols of the digits (i.e. 0-9) 
- b) The length of the string should be at least 8 symbols 
- c) The string must contain at least 1 uppercase symbol
If all the conditions above are satisfied, you need to print the message ‘The password is valid’, otherwise print the message with information on which conditions were broken, separated by a comma.

**Example**: 
- Input: password = ‘0This_is_a_valid_password_0’ 
- Output: ‘The password is valid’ 

- Input: password = ‘short1’ 
- Output: ‘The given password is too short, digits condition is broken, no uppercase symbol was given’

**Hints**:

How we count the specific element in the List or str?

``` python
    if i.isdigit():

    if i.isupper(): ...
     
```

````python
password = input()
broken_conditions = []
digits_count = 0
upper_count = 0

if len(password) < 8:
    broken_conditions.append("Password is too short")

for char in password:
    if char.isupper():
        upper_count += 1
    if char.isdigit():
        digits_count += 1

if digits_count <2:
    broken_conditions.append("digits condition is broken")

if upper_count == 0:
    broken_conditions.append("no uppercase symbol was given")

if len(broken_conditions) == 0:
    print("The password is valid")
else:
    merged_text = ", ".join(broken_conditions)
    print(f"The password is invalid because: {merged_text}")
````

### Task 3: Count Vowels in a String
Given a string `text`, count how many vowels (`a, e, i, o, u`) are present in it.  
The program should print the number of vowels.

**Example:**
- Input: `"Hello World"`
- Output: `3`


````python
text = input()
vowels = ["a", "e", "i", "o", "u"]
count = 0

for char in text:
    if char.lower() in vowels:
        count += 1

print(count)
````

### Task 4: Student Grade Evaluation
Given a list of student scores, check the grade according to these rules:
- If score ≥ 90 → Grade A
- If 75 ≤ score < 90 → Grade B
- If 60 ≤ score < 75 → Grade C
- If score < 60 → Grade F

Additionally, print:
- `"Excellent"` if Grade A
- `"Good"` if Grade B
- `"Pass"` if Grade C
- `"Fail"` if Grade F

**Example:**
- Input: [95, 82, 67, 45]  
- Output:  

Score: 95 → Grade A (Excellent)

Score: 82 → Grade B (Good)

Score: 67 → Grade C (Pass)

Score: 45 → Grade F (Fail)


````python
scores = [95, 82, 67, 45]

for score in scores:
    if score >= 90:
        print("Score: ", score, "-> Grade: A (Excellent)")
    elif score >= 75:
        print("Score: ", score, "-> Grade: B (Good)")
    elif score >= 60:
        print("Score: ", score, "-> Grade: C (Pass)")
    else:
        print("Score: ", score, "-> Grade: F (Fail)")
````
