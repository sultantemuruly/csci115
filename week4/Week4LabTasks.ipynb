##  Problem 1: While Loop and Summation

Using a **while loop**, repeatedly prompt the user for an integer:

1. Keep a running total of all numbers entered.
2. If the user enters a negative number, stop asking for more numbers (i.e., break the loop).
3. Print the total sum of all non-negative integers entered.

# Expected output for Problem 1 (example scenario):

User enters: 5, 4, 10, -1
Total sum of non-negative integers: 19

(You may have a different set of inputs, but the logic should match.)

````python
nums = []
while True:
    num = int(input())
    if num < 0:
        break
    nums.append(num)

sum = 0

for num in nums:
    if num >= 0:
        sum += num

print('Sum of non-negative numbers:', sum)
````

## Problem 2: Weekly Health Habits Analysis
Each index represents the same day (Monday → Sunday).

````python
steps = [4500, 8200, 10300, 6700, 12000, 3000, 9000]
water_liters = [1.2, 2.0, 2.5, 1.8, 3.0, 1.0, 2.2]
sleep_hours = [6, 7.5, 8, 5, 8.5, 6.5, 7]
````

### Task 2.1: Count Active Days 
A day is **active** if the number of steps is **8000 or more**.

**Your task:**
- Count how many active days there are
- Print the result

````python
count = 0

for step in steps:
    if step >= 8000:
        count += 1

print('Number of active days:', count)
````

### Task 2.2: Daily Health Status 
For each day, determine the health status:

- **Good day** if:
  - steps ≥ 8000
  - water ≥ 2 liters
  - sleep ≥ 7 hours

- **Okay day** if **at least one** condition is met
- **Poor day** if **none** of the conditions are met

Print the result for each day (Day 1, Day 2, ...).

````python
day_type = {}

for i in range(len(steps)):
    count = 0
    if steps[i] >= 8000:
        count += 1
    if water_liters[i] >= 2:
        count += 1
    if sleep_hours[i] >= 7:
        count += 1
    if count == 3:
        key = 'Day ' + str(i + 1)
        day_type[key] = 'Good'
    elif count >= 1 and count <= 2:
        key = 'Day ' + str(i + 1)
        day_type[key] = 'Okay'
    else:
        key = 'Day ' + str(i + 1)
        day_type[key] = 'Poor'

for day, status in day_type.items():
    print(f"{day}: {status}")
````

### Task 2.3: Find the Worst Day 
The **worst day** is the day with the **minimum number of steps**.

**Your task:**
- Find the day number (1–7)
- Print the day and the number of steps

````python
minimum_steps = steps[0]
day_index = 0

for i in range(len(steps)):
    if steps[i] < minimum_steps:
        minimum_steps = steps[i]
        day_index = i

print("Day:", day_index + 1, "with minimum steps:", minimum_steps)
````

### Task 2.4: Motivation Message 
For each day:
- steps < 5000 → `Try to move more tomorrow!`
- 5000 ≤ steps < 8000 → `Good effort, keep going!`
- steps ≥ 8000 → `Great job!`

````python
for steps in steps:
    if steps < 5000:
        print("Try to move more tomorrow!")
    elif steps < 8000 and steps >= 5000:
        print("Good effort, keep going!")
    else:
        print("Great job!")
````

# Problem 3: Filtering Words with Conditionals

Given a list of words, use a **for loop** and **if** statement to:

Print both lists.
1. Create a new list called `consonant_words` that contains only the words that start with a consonant. Consonants are all letters except: a, e, i, o, u
2. Create another list called `vowel_words` that contains words that start with a vowel (a, e, i, o, u).
3. Print both lists.

Example list:
```
words = ["apple", "banana", "orange", "umbrella", "cake", "egg"]
```

# Expected output for Problem 3:

consonant_words: ['banana', 'cake']
vowel_words: ['apple', 'orange', 'umbrella', 'egg']

````python
vowels = ['a', 'e', 'i', 'o', 'u']
words = ["apple", "banana", "orange", "umbrella", "cake", "egg"]

consonant_words = []
vowel_words = []

for word in words:
    if word[0].lower() in vowels:
        vowel_words.append(word)
    else:
        consonant_words.append(word)
        
print("consonant_words:", consonant_words)
print("vowel_words:", vowel_words)
````

# Problem 4: Nested Loops (Multiplication Table)

Use **nested for loops** to create a multiplication table from 1 to 5 (or 1 to 10 for a bigger table):

1. Print the results in a formatted manner (e.g., each row on its own line).
2. Optional: Align the columns.

Example: A 1-to-5 multiplication table might look like:

```
1  2  3  4  5
2  4  6  8  10
3  6  9  12 15
4  8  12 16 20
5  10 15 20 25
```

````python
for i in range(1, 6):
    row = ''
    for j in range(1, 6):
        row += f'{i*j:<2} '
    print(row)
````

````python
matrix = []

for i in range(1, 6):
    row = []
    for j in range(1, 6):
        row.append(i * j)
    matrix.append(row)


for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        print(matrix[i][j], end=' ')
    print(' ')
````

## Task 5. 1:

While printing the output table from Problem 4:

- if a value is even, print it normally

- if a value is odd, print * instead

````python
# #  Expected result:


# *   2   *   4   *
# 2   4   6   8   10
# *   6   *   12  *
# 4   8   12  16  20
# *   10  *   20  *
````

````python
for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if matrix[i][j] % 2 != 0:
            matrix[i][j] = '*'

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        print(matrix[i][j], end=' ')
    print(' ')
````

## Task 5.2:

Using the output table from Problem 4 print only:

- the diagonal elements (e.g., 1×1, 2×2, …)

- replace all other values with -

````python
for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if i != j:
            matrix[i][j] = '-'

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        print(matrix[i][j], end=' ')
    print(' ')
````
