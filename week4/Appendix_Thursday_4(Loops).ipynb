## More examples
- Carefully check if we have multiple conditions: 
```python
if 3 in my_list and 'H' in my_str:
```
This can be interpreted as :
```python
if (3 in my_list) and ('H' in my_str):
```
This will return "True" only if both conditions are "True"

````python
# Define some variables
a = 10
b = 20
c = 30

# List and string for membership testing
my_list = [1, 2, 3, 4, 5]
my_str = "Hello"

# Using Comparison and Logical Operators
if a < b and b < c:
    print("Both conditions are True: a < b and b < c")

# Using Logical 'or' with Comparison
if a > b or a <= 10:
    print("At least one condition is True: a > b or a <= 10")

# Using Logical 'not' with Comparison
if not a == b:
    print("a and b are not equal")

# Using Logical and Membership operators
if 3 in my_list and 'H' in my_str:
    print("3 is in my_list and 'H' is in my_str")

# Using Logical or Membership operators
if 3 not in my_list or 'e' in my_str:
    print("Either 3 is not in my_list or 'e' is in my_str")

# Using all types of operators
if (a < b and 'H' in my_str) or (3 in my_list and c >= b):
    print("Complex condition satisfied")

# Nested if statements with various operators
if a < b:
    if 3 in my_list:
        if 'l' in my_str and c > a:
            print("All nested conditions are True")
````

# Loops in Python

Loops are an essential concept in programming that allow us to execute a block of code multiple times. Python provides two primary types of loops:

- **for loop**: Iterates over a sequence (such as a list, tuple, dictionary, string, or range). 
1) no more elements to iterate, 
2) break statement

```python
str1 = "hello everyone"
list1 = ["h", "e", "l"]
dict1 = {"a:1", "b:2"}

for i in str1:
    print(i)
    #or str1[index]
```


- **while loop**: Repeats as long as a given condition is `True`.
1) condition becomes False
2) break statement


#### For loops
```python
for i in sequences:
    print(i)
    
for i in range(10):
    print(i)
for i in [0, 1, 2, 3]:
    print(i)
for i in [a, b, c, d]:
    print(i)
    
str1 = 'hello'    
for i in str1:
    print(i)

for number in range(10):
    if number == 5:
        break  # Stops the loop when number reaches 5
```

#### while loops
```python

    loop body
```

### Make this code with for / while loops

````python
data = "apple: алма, school: мектеп, car: көлік, house: үй, computer: компьютер, phone: телефон, book: кітап, tree: ағаш, river: өзен, mountain: тау"
pairs = data.split(", ")
print(pairs)
eng = []
kaz = []
eng.append(pairs[0].split(":")[0])
kaz.append(pairs[0].split(":")[1])

eng.append(pairs[1].split(":")[0])
kaz.append(pairs[1].split(":")[1])

eng.append(pairs[2].split(":")[0])
kaz.append(pairs[2].split(":")[1])

eng.append(pairs[3].split(":")[0])
kaz.append(pairs[3].split(":")[1])

eng.append(pairs[4].split(":")[0])
kaz.append(pairs[4].split(":")[1])

eng_idx = eng.index("school")
print(kaz[eng_idx])
````

````python
for pair in pairs:
    e, k = pair.split(": ")
    eng.append(e)
    kaz.append(k)

for i, pair in enumerate(pairs): # return pairs of index and value.
    e, k = pair.split(": ")
    eng.append(e)
    kaz.append(k)

for i in range(len(pairs)):
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
````

````python
i = 0
while True:
    if i >= len(pairs):  # Exit condition
        break
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
    i += 1
    
i = 0
while i != -1:
    if i >= len(pairs):  # Stop condition
        break
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
    i += 1

i = 0
while i != len(pairs):  # Alternative condition
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
    i += 1

i = 0
while i in range(len(pairs)):  # Condition based on range
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
    i += 1

i = 0
while pairs[i:]:  # Directly using the list slice as a condition
    e, k = pairs[i].split(": ")
    eng.append(e)
    kaz.append(k)
    i += 1
````
