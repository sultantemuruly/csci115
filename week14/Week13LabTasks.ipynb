5. Use broadcasting to multiply them and produce a 10×10 multiplication table. Store the result in a variable called `result`. No loops allowed!
   

```python
import numpy as np

rows = np.arange(1, 11)
rows = rows.reshape(1, 10)
cols = np.arange(1, 11)
cols = cols.reshape(10, 1)
result = rows * cols
result
```

6. Once you have `rows`, `cols`, and `result`, run the cell below without any modification. If your code is correct, you will see a valid equation and a multiplication table.

```python
from IPython.display import display, Math
import random

def to_latex(arr, highlight=False):
    n, m = np.atleast_2d(arr).shape
    if highlight:
        r1, c1 = random.randint(0, n-2), random.randint(0, m-2)
        r2, c2 = random.randint(r1+1, min(r1+5, n)), random.randint(c1+1, min(c1+5, m))
    lines = []
    for i, row in enumerate(np.atleast_2d(arr)):
        cells = [rf"\color{{#DC143C}}{{{v}}}" if highlight and r1<=i<r2 and c1<=j<c2 else str(v)
                 for j, v in enumerate(row)]
        lines.append(" & ".join(cells))
    return r"\begin{bmatrix}" + r"\\".join(lines) + r"\end{bmatrix}"

display(Math(to_latex(rows) + r"\times" + to_latex(cols) + "=" + to_latex(result, highlight=True)))
```

### 3.1 Display the image

Run the cell below to create the array and display it. You should see the letters "NU".

```python
# Run this cell. DO NOT MODIFY

import numpy as np

art = np.zeros((12, 12), dtype=int)

art[1:11, 0] = 1
art[1:11, 5] = 1
for i in range(10):
    row = 1 + i
    col = int(0.5 + i * 5 / 9)
    art[row, col] = 1

art[1:10, 7] = 1
art[1:10, 11] = 1
art[10, 8:11] = 1

def show(arr):
    """Display a 2D array of 0s and 1s as # and . characters."""
    for row in arr:
        print(" ".join(["#" if val else "." for val in row]))

show(art)
```

## Task 4. Distance Between Cities (broadcasting + reshape + argmin)

You are given the coordinates (latitude and longitude) of 5 cities in Kazakhstan. Your goal is to compute the straight-line distance between every pair of cities using NumPy, without any loops.

The Euclidean distance between two points $(lat_1, lon_1)$ and $(lat_2, lon_2)$ is:

$$d = \sqrt{(lat_1 - lat_2)^2 + (lon_1 - lon_2)^2}$$

**Note:** This is a simplified distance using coordinates as plain numbers, not the real geographic distance. It is enough for comparison purposes.

### 4.1 Set up the data

Run the cell below to create the arrays. Do not modify this code.

```python
# Run this cell. DO NOT MODIFY

names = ["Astana", "Almaty", "Shymkent", "Aktobe", "Karaganda"]
lat = np.array([51.17, 43.14, 42.19, 50.30, 49.80])
lon = np.array([71.43, 76.57, 69.35, 57.21, 73.10])
```

### 4.2 Build the distance matrix

Compute a 5×5 distance matrix where entry `[i, j]` is the distance between city `i` and city `j`. Use broadcasting, no loops allowed!

**Step-by-step guide:**

1. Reshape `lat` into a column vector of shape `(5, 1)`. Subtract the original `lat` of shape `(5,)` from it. Broadcasting will produce a `(5, 5)` matrix of latitude differences. Store it in `dlat`.

2. Do the same for `lon` → store in `dlon`.

3. Apply the distance formula: `dist = np.sqrt(dlat**2 + dlon**2)`.

4. Print the result using the `show_matrix()` function provided below.

**Hint:** This is the same broadcasting idea as Task 1: a column vector times (or minus) a row vector gives a 2D matrix.

```python
# Run this cell. DO NOT MODIFY
# Helper function to display the distance matrix with city names

def show_matrix(matrix, names):
    header = "           " + "  ".join([f"{n:>10}" for n in names])
    print(header)
    for i in range(len(names)):
        row = f"{names[i]:>10} " + "  ".join([f"{matrix[i,j]:10.2f}" for j in range(len(names))])
        print(row)
```

```python
dlat = lat.reshape(5, 1) - lat
dlon = lon.reshape(5, 1) - lon
dist = np.sqrt(dlat**2 + dlon**2)

show_matrix(dist, names)
```

### 4.3 Find the nearest neighbor

For each city, find which other city is closest to it.

**Step-by-step guide:**

1. First, set the diagonal of `dist` to `np.inf`. This prevents a city from being its own nearest neighbor. You can do this with: `dist[np.arange(5), np.arange(5)] = np.inf`

2. Use `np.argmin(dist, axis=1)` to find the index of the smallest value in each row.

3. Print the nearest neighbor for each city.

**Expected output:**

Astana -> Karaganda <br>
Almaty -> Shymkent <br>
Shymkent -> Almaty <br>
Aktobe -> Astana <br>
Karaganda -> Astana <br>


```python
dist[np.arange(5), np.arange(5)] = np.inf
indexes = np.argmin(dist, axis=1)
for i in range(5):
    print(f"{names[i]} -> {names[indexes[i]]}")
```

### 4.4 Find the farthest pair

Find the two cities that are farthest apart from each other.

**Step-by-step guide:**

1. Reset the diagonal back to 0: `dist[np.arange(5), np.arange(5)] = 0`

2. Find the maximum distance using `dist.max()`.

3. Use `np.where(dist == dist.max())` to find the row and column indices of that maximum value.

4. Print the names of the two cities and the distance between them.

```python
dist[np.arange(5), np.arange(5)] = 0
indexes = np.argmax(dist, axis=1)
for i in range(5):
    print(f"{names[i]} -> {names[indexes[i]]}")
```
