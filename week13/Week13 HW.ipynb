1. Make a class called **Circle** with an attribute radius. Add two methods: _area()_ and _circumference()_, that calculate and return the circle’s area and circumference. Create an instance of Circle and display the results of both methods.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def circumference(self):
        return 2 * 3.14159 * self.radius    
```

```python
circle = Circle(5)
print(circle.area())          
print(circle.circumference()) 
```

2. Make a class called **Book** with attributes _title_, _author_, and _copies_available_.
Add a method _describe_book()_ that prints the title, author, and number of copies available.
Add a method _borrow_book()_ that decreases copies_available by 1 each time it is called, but prints a message like “No copies left!” if there are none available.
Add another method _return_book()_ that increases the number of available copies by 1.
Create an instance of the Book class and demonstrate the use of all three methods.

```python
class Book:
    def __init__(self, title, author, copies_available):
        self.title = title
        self.author = author
        self.copies_available = copies_available
    
    def describe_book(self):
        return f"{self.title} | {self.author} | ({self.copies_available} copies available)"
    
    def borrow_book(self):
        if self.copies_available > 0:
            self.copies_available -= 1
            print("Borrowed")
        else:
            print("No copies left!")
    
    def return_book(self):
        self.copies_available += 1
        print("Returned")
```

```python
book = Book("Name", "Author", 3)
print(book.describe_book())
```

```python
book.borrow_book()
book.borrow_book()
book.borrow_book()
book.borrow_book()
```

```python
book.return_book()
book.return_book()
```
