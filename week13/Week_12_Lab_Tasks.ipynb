## Task 1 — Car Class

Define a class `Car` with attributes `brand`, `model`, and `year`.

Add a method `get_info()` that returns `"2020 Toyota Corolla"`.

Add a setter method `set_year(y)` that updates the year. If `y` is less than `1900`, print `"Invalid year"` and do not update.

Save the class to `car.py` using %%writefile and import it.

```
Expected output:
2020 Toyota Corolla
Invalid year
2021 Toyota Corolla
```

````python
%%writefile car.py
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def get_info(self):
        return f"{self.year} {self.brand} {self.model}"
    
    def set_year(self, new_year):
        if(new_year < 1900): 
            print("Invalid year")
            return
        self.year = new_year
````

````python
from car import Car

toyota = Car("Toyota", "Corolla", 2020)
print(toyota.get_info())
toyota.set_year(1800)
````

## Task 2 — Student Class

Create a `Student` class with attributes `name`, `student_id`, and `grades` — a dictionary `{subject: grade}` that starts empty.

Add the following methods:
- `add_grade(subject, grade)` — adds the grade. If the grade is outside 0–100, print `"Invalid grade"` and do not add it.
- `get_average()` — returns the average of all grades, or `0.0` if none exist.
- `get_info()` — returns `"Alice (ID: S-01) | Average: 85.67"`

```
Expected output:
Invalid grade
Alice (ID: S-01) | Average: 85.67
```

````python
class Student():
    def __init__(self, name, student_id, grades):
        self.name = name
        self.student_id = student_id
        self.grades = grades
    
    def add_grade(self, course, grade):
        self.grades[course] = grade
    
    def get_average_grade(self):
        return sum(self.grades.values()) / len(self.grades)

    def get_info(self):
        return f"{self.name} (ID: {self.student_id}) | Average: {self.get_average_grade():.2f}"
````

````python
alice = Student('Alice', 'S-01', {'Math': 100, 'Economics': 96})
alice.add_grade('History', 97)
print(alice.get_info())
````

## Task 3 — Animal and Dog

Define a base class `Animal` with attributes `name` and `age`, a method `get_info()` that returns `"Name: Buddy, Age: 3"`, and a method `speak()` that prints `"..."`.

Define a child class `Dog(Animal)` that adds a `breed` attribute, overrides `get_info()` to produce `"Name: Buddy, Age: 3, Breed: Labrador"`, and overrides `speak()` to print `"Woof!"`.

Save both classes to `animals.py` and import with `from animals import Animal, Dog`.

```
Expected output:
Name: Buddy, Age: 3
...
Name: Buddy, Age: 3, Breed: Labrador
Woof!
```

````python
%%writefile animals.py
class Animal():
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def get_info(self):
        return f"Name: {self.name}, Age: {self.age}"
    
    def speak(self):
        return "..."

class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed

    def get_info(self):
        return f"Name: {self.name}, Age: {self.age}, Breed: {self.breed}"
    
    def speak(self):
        return "Woof!"
````

````python
from animals import Animal, Dog

animal = Animal("Buddy", 5)
print(animal.get_info())
print(animal.speak())

dog = Dog("Buddy", 5, "Labrador")
print(dog.get_info())
print(dog.speak())
````

## Task 4 — Pet and Owner

Define a class `Owner` with attributes `name` and `phone`, and a method `get_info()` returning `"Alice | 555-1234"`.

Define a class `Pet` with attributes `pet_name`, `species`, and `owner` — an instance of `Owner`. Add `get_info()` that returns `"Whiskers (cat) — Owner: Alice | 555-1234"`.

Then demonstrate the three copying behaviors using `import copy`. For each step, add a comment explaining why the output is what it is.

1. `pet2 = pet1`. Change `pet2.pet_name`. Print both names.
2. `pet3 = copy.copy(pet1)`. Change `pet3.pet_name`, then `pet3.owner.name`. Print both pet names, then both owner names.
3. `pet4 = copy.deepcopy(pet1)`. Change `pet4.owner.name`. Print both owner names.

```
Expected output:
Whiskers (cat) — Owner: Alice | 555-1234
--- Alias ---
pet1: Shadow   pet2: Shadow
--- Shallow copy ---
pet1: Shadow   pet3: Luna
pet1 owner: Bob   pet3 owner: Bob
--- Deep copy ---
pet1 owner: Bob   pet4 owner: Carol
```

````python
class Owner():
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    def get_info(self):
        return f"{self.name} | {self.phone}"

class Pet():
    def __init__(self, pet_name, species, owner):
        self.pet_name = pet_name
        self.species = species
        self.owner = owner
    
    def get_info(self):
        return f"{self.pet_name} ({self.species}) - Owner: {self.owner.get_info()}"
````

````python
alice = Owner("Alice", "555-1234")
pet1 = Pet("Whiskers", "Cat", alice)

print(alice.get_info())
print(pet1.get_info())
````

````python
pet2 = pet1
pet2.pet_name = "Not Whiskers"

print(pet1.get_info())
print(pet2.get_info()) 

# both names changed because pet1 and pet2 reference the same object in memory.
````

````python
import copy

pet3 = copy.copy(pet1)
pet3.pet_name = "Not Whiskers 2"
pet3.owner.name = "Not Alice 2"

print(pet3.get_info())
print(pet1.get_info()) 

# pet3 is a shallow copy of pet1, so pet3.pet_name is different but pet3.owner is the same object as pet1.owner. Therefore, changing pet3.owner.name also changes pet1.owner.name.
# shallow copy – does not copy embedded objects; copies references 
````

````python
pet4 = copy.deepcopy(pet1)
pet4.owner.name = "Not Alice 3"

print(pet4.get_info())
print(pet1.get_info())
 
# pet1 and pet4 are completely independent objects, so changing pet4.owner.name does not affect pet1.owner.name.
````
