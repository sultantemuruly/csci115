### Question 3
Given three lists with student information of student_ids, student_subjects, student_grades. 
```python
student_ids = [ 'ID9012', 'ID4212', 'ID7809' ]
student_subjects = [ 'Programming Fundamentals', 'Calculus', 'Algebra' ]
student_grades = [ 78, 90, 75 ]
```

Create a nested dictionary by combining entries from these lists, such as for each student entry with the given student id, the nested dictionary contains a student subject with the grade information for that subject.

#### Expected output: 
```python
{'ID9012': {'Programming Fundamentals':78}, 'ID4212': {'Calculus': 90},'ID7809': {'Algebra': 75}}
```

````python
student_ids = [ 'ID9012', 'ID4212', 'ID7809' ]
student_subjects = [ 'Programming Fundamentals', 'Calculus', 'Algebra' ]
student_grades = [ 78, 90, 75 ]

output = {}

for i in range(len(student_ids)):
    output[student_ids[i]] = {
        'subject': student_subjects[i],
        'grade': student_grades[i]
    }

print(output)
````

### Question 4

Given a list of colors. 
```python
colors = [ '#C2345C', '#B71243', '#76121A' , '#C2345C', '#76121A', '#76121A' ]

```
Some colors in this list might be repeated several times. Create 
a dictionary with color keys and the frequency of their appearance in the colors list.
Output entries of the dictionary in increasing order of their frequency.

#### Expected output: 
```python
>>> Colors dictionary
{'#C2345C' : 2, '#B71243': 1, '76121A': 3}
Color #B71243 appear 1 times in the list
Color #C2345C appear 2 times in the list
Color #76121A appear 3 times in the list
 ```
 #### Hint: 
 
You can find unique elements using 'unique' funcion in 'numpy' library. 
- import an external library to your workspace: "import numpy", then find the unique function within the library, i.e., numpy.unique

You can acquire the frequencies for the specific element using 'count' function.   


```python
import numpy
a = [1, 2, 3, 4, 1, 2, 3, 2, 2]
unique_a = numpy.unique(a)  # this will return unique elements in 'a'
a.count(unique_a[0])  # you can count the specific element in the a, same as 'a.count(1)'
print(f'Element {unique_a[0]} has been occurred {a.count(unique_a[0])} times in the variable a')
```

````python
import numpy
colors = [ '#C2345C', '#B71243', '#76121A' , '#C2345C', '#76121A', '#76121A' ]
unique_colors = numpy.unique(colors)

output = {}

for color in unique_colors:
    count = colors.count(color)
    output[str(color)] = count
    
print(output)
````
