## Task 1 — Basic List Operations
Create a list of integers entered by the user (5 numbers).

Output:

````python
my_list = []

num1 = int(input("Enter a number: "))
my_list.append(num1)

num2 = int(input("Enter a number: "))
my_list.append(num2)

num3 = int(input("Enter a number: "))
my_list.append(num3)

num4 = int(input("Enter a number: "))
my_list.append(num4)

num5 = int(input("Enter a number: "))
my_list.append(num5)

print("List:", my_list)
print("First element:", my_list[0])
print("Last element:", my_list[-1])

sum_1 = sum(my_list)
print("Sum:", sum_1)

average = sum_1 / 5
print("Average:", average)
````

## Task 2 — List Filtering
Given a predefined list 
```python
numbers = [3, 8, 1, 9, 4, 7, 2]
```
Create a new list containing only even numbers and print it.


````python
numbers = [3, 8, 1, 9, 4, 7, 2]
even_numbers = []

num_1 = numbers[0]
if num_1 %2 == 0:
    even_numbers.append(num_1)

num_2 = numbers[1]
if num_2 %2 == 0:
    even_numbers.append(num_2)

num_3 = numbers[2]
if num_3 %2 == 0:
    even_numbers.append(num_3)

num_4 = numbers[3]
if num_4 %2 == 0:
    even_numbers.append(num_4)

num_5 = numbers[4]
if num_5 %2 == 0:
    even_numbers.append(num_5)
    
num_6 = numbers[5]
if num_6 %2 == 0:
    even_numbers.append(num_6)
    
num_7 = numbers[6]
if num_7 %2 == 0:
    even_numbers.append(num_7)
    
print("Even numbers:", even_numbers)
````

## Task 6 — Dictionary Creation
Create a dictionary storing student grades:
```python
grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78
}
```

1. Print Bob’s grade
2. Add a new student "Diana" with grade 90
3. Print the updated dictionary

````python
grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78
}

print("Bob's grade:", grades["Bob"])
grades["Diana"] = 90
print("Updated dictionary:", grades)
````

## Task 7 — Dictionary Iteration
Using the dictionary from Task 6, print each student and grade in the format:

1) Alice -> 85
2) Bob -> 92
3) Charlie -> 78
4) Diana -> 90

````python
keys = list(grades.keys())

print(keys[0], '->', grades[keys[0]])
print(keys[1], '->', grades[keys[1]])
print(keys[2], '->', grades[keys[2]])
print(keys[3], '->', grades[keys[3]])
````

## Bonus tasks #4 — Dictionary Keys and Values
Create the dictionary:
```python
floors = {
    '1F': 'Fruit',
    '2F': 'Meat',
    '3F': 'Electronics'
}
```
1. Print the value for key '2F'
2. Add a new key '4F' with value 'Clothes'
3. Change '1F' to 'Books'
4. Print the updated dictionary
5. Create a list of all keys of `floors`
6. Create a list of all values of `floors`
7. Print both lists

````python
floors = {
    '1F': 'Fruit',
    '2F': 'Meat',
    '3F': 'Electronics'
}

print(floors['2F'])

floors['4F'] = 'Clothes'
floors['Books'] = floors.pop('1F')
print(floors)

list_keys = list(floors.keys())
list_values = list(floors.values())
print(list_keys)
print(list_values)
````
