### Task 1.1: Read specific chunks and track position

Open `data.txt` in read mode and perform the following steps **in order**:

1. Print the current position using `tell()` → should be `0`
2. Read and print the first 5 characters → should be `ABCDE`
3. Print the current position using `tell()`
4. Read and print the next 5 characters → should be `FGHIJ`
5. Print the current position using `tell()`
6. Use `seek(0)` to go back to the beginning
7. Read and print the first 3 characters again → should be `ABC`

**Expected output:**

## Task 3: Word Frequency from a File [Dictionaries + File I/O]

This task combines file reading with dictionaries and functions.

### Task 3.1: Build a word frequency dictionary

Write a function called `word_frequency(filename)` that:

1. Opens the given file and reads its contents.
2. Converts all text to **lowercase**.
3. Removes the following punctuation characters: `. , ! ? ( ) : ;`  
   (Hint: use a loop with `.replace()` for each punctuation character)
4. Splits the cleaned text into words.
5. Counts the frequency of each word using a **dictionary**.
6. **Returns** the dictionary.

Then call the function on `sample.txt` and print the result.

**Expected output (partial):**
```
python -> 5
was -> 2
created -> 1
by -> 1
...
```

**Requirements:**
- The function must use `with` context manager
- Use a `dict` to store word frequencies
- Use a `for` loop to iterate over words
- Use `print()` with `sep` parameter to display results (e.g., `print(word, count, sep=' -> ')`)

````python
# Write your code here
# Run this cell to create the sample file — DO NOT MODIFY
sample_text = """Python was created by Guido van Rossum.
It was first released in 1991.
Python emphasizes code readability.
The language supports multiple programming paradigms.
Python is widely used in web development and data science.
Many beginners choose Python as their first language.
The Python community is large and supportive."""

with open("sample.txt", "w") as f:
    f.write(sample_text)
````

````python
def word_frequency(filename):
    symbols = ['.', ',', '!', '?', ';', ':', '(', ')']
    frequency = {}
    
    with open(filename, 'r') as f:
        for line in f:
            words = line.split()
            for symbol in symbols:
                words = [word.replace(symbol, '').lower() for word in words]
            for word in words:
                if word in frequency:
                    frequency[word] += 1
                else:
                    frequency[word] = 1
    return frequency
````

````python
frequency = word_frequency("sample.txt")

for word, count in frequency.items():
    print(f"{word} -> {count}")
````

### Task 3.2: Find the top 3 most frequent words

Using the dictionary returned from `word_frequency()`, find and print the **3 most frequent words** and their counts.

**Expected output:**
```
Top 3 most frequent words:
1. python -> 5
2. the -> 3
3. is -> 2
```

**Hint:** You can use `sorted()` with a `key` parameter to sort dictionary items by value:
```python
sorted(my_dict.items(), key=lambda x: x[1], reverse=True)
```

**Requirements:**
- Use the dictionary from Task 6.1 (don't re-read the file)
- Use `sorted()` to sort by frequency
- Print only the top 3

````python
list_frequency = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

i = 0
while i < 3:
    word, count = list_frequency[i]
    print(f"{i+1}. {word} -> {count}")
    i+=1
````

### Task 3.3: Save the frequency table to a file

Write the full word frequency table (sorted alphabetically by word) to a file called `word_freq.txt`.

**Expected content in `word_freq.txt`:**
```
and -> 2
as -> 1
beginners -> 1
by -> 1
choose -> 1
...
```

After writing, read the file back and print its contents to confirm.

**Requirements:**
- Sort the dictionary by keys (alphabetical order)
- Use `print()` with `sep` and `file` parameters to write each line
- Use `with` context manager

````python
with open('word_freq.txt', 'w') as f:
    keys = list(frequency.keys())
    sorted_keys = sorted(keys)
    for key in sorted_keys:
        f.write(f"{key} -> {frequency[key]}\n")

with open('word_freq.txt', 'r') as f:
    print(f.read())
````
