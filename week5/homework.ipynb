```python
def absolute_difference(a, b):
    c = a-b
    
    if c < 0: c *= -1
    
    return int(c)
```

```python
res = absolute_difference(3, 1)
print(res)
```

```python
def count_vowels(s):
    vowels = ['a', 'e', 'i', 'o', 'u', 'y']
    count = 0
    for letter in s:
        if letter.lower() in vowels:
            count += 1
    return count
```

```python
res = count_vowels('LettEr')
print(res)
```

```python
def average(numbers):
    numbers_sum = 0
    if (len(numbers) == 0):
        return 0.0
    for num in numbers:
        numbers_sum += num
    ave = float(numbers_sum)/(len(numbers))
    ave = round(ave, 2)
    return ave
```

```python
res = average([10, 10, 12, 12, 3, 12, 13, 169, 129, 120.2])
print(res)
```

```python
def is_even(num):
    if num % 2 == 0:
        return True
    return False
```

```python
res = is_even(3)
print(res)
```

```python
def classify_grade(grade):
    if grade >= 90:
        return 'A'
    elif grade >= 80:
        return 'B'
    elif grade >= 70:
        return 'C'
    elif grade >= 60:
        return 'D'
    else:
        return 'F'
```

```python
res = classify_grade(90)
print(res)
```
