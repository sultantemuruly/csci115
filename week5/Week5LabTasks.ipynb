## Task 1 — Circle calculator

1. Write two functions:
   - `circle_length(r)` — takes the radius and returns the circumference of a circle: $L = 2\pi r$
   - `circle_area(r)` — takes the radius and returns the area of a circle: $S = \pi r^2$

   For $\pi$, you can use an approximate value of `3.1415`.

2. Then ask the user to input a radius, call both functions on the input value, and print the results rounded to **2 decimal places**.

**Example (user enters `5`):**
```
Enter radius: 5
Circumference: 31.42
Area: 78.54
```


````python
def circle_length(r):
    return round(2 * 3.1415 * r, 2)

def circle_area(r):
    return round(3.1415 * r**2, 2)

radius = int(input('Enter radius'))
circumference = circle_length(radius)
area = circle_area(radius)

print(f'Circumference: {circumference}')
print(f'Area: {area}')
````

## Task 2 — Greeting with a language

1. Write a function called `greet` that takes two parameters:

   - `language` (string with default value `"en"`)

&nbsp;

   - `"kk"` → `"Сәлем, {name}!"`

**Example (user enters `Alice`):**

## Task 3 — Temperature converter (two-way)

1. Write a function called `convert_temp` that takes two parameters:
   - `value` — a number (the temperature)
   - `to_unit` — a string, either `"F"` or `"C"` (with default value `"F"`)

&nbsp;
   The function should convert and return the result rounded to **1 decimal place** (this can be done by applying the function `round(x, 1)`, where `x` is the value to be rounded):
   - If `to_unit` is `"F"`, convert Celsius → Fahrenheit: $F = C \times \frac{9}{5} + 32$
   - If `to_unit` is `"C"`, convert Fahrenheit → Celsius: $C = (F - 32) \times \frac{5}{9}$

&nbsp;
2. Ask the user to input a temperature in Celsius, then print both the Celsius → Fahrenheit and Fahrenheit → Celsius conversions.

**Example (user enters `100`):**
```
40°C -> 104.0°F
40°F -> 4.4°C
```


````python
def convert_temp(temp, to_unit):
    if to_unit == 'F':
        return round(temp * 9/5 + 32, 1)
    else:
        return round((temp - 32) * 5/9, 1)

temp = float(input())

temp_f = convert_temp(temp, 'F')
temp_c = convert_temp(temp, 'C')

print(f'{temp}°C -> {temp_f}°F')
print(f'{temp}°F -> {temp_c}°C')
````

## Task 4 — Password strength checker

1. Write a function called `check_password` that takes a password string and returns a strength rating:
   - `"Weak"` — if the password is shorter than 6 characters
   - `"Medium"` — if it is 6–9 characters long
   - `"Strong"` — if it is 10 or more characters long **and** contains at least one digit

&nbsp;
2. Using the function and the given list, print each password and its strength.

**Expected output:**
```
abc -> Weak
qwerty123 -> Medium
strongpass9 -> Strong
longpassword -> Medium
```

````python
passwords = ["abc", "qwerty123", "strongpass9", "longpassword"]

def check_password(password):
    if len(password) < 6:
        return 'Weak'
    elif len(password) < 10:
        return 'Medium'
    else:
        for letter in password:
            if letter in '0123456789':
                return 'Strong'
        return 'Medium'

for password in passwords:
    strength = check_password(password)
    print(f'{password} -> {strength}')
````

You are given the following dictionary of students and their scores:

    "Charlie": [95, 100, 98],
    "Diana": [40, 35, 50]

### Task 5.1 — Average calculator

Write a function `average(scores)` that takes a list of numbers and returns the average (rounded to 1 decimal).

### Task 5.2 — Letter grade
Write a function `letter_grade(avg)` that takes an average score and returns a letter grade:

- 90 and above → "A"
- 80–89 → "B"
- 70–79 → "C"
- 60–69 → "D"
- below 60 → "F"

### Task 5.3 — Print the report
Using both functions, loop through the `students` dictionary and print a formatted report.

Expected output:
```
Alice    | Avg: 86.3 | Grade: B
Bob      | Avg: 62.7 | Grade: D
Charlie  | Avg: 97.7 | Grade: A
Diana    | Avg: 41.7 | Grade: F
```


````python
students = {
    "Alice": [88, 92, 79],
    "Bob": [55, 63, 70],
    "Charlie": [95, 100, 98],
    "Diana": [40, 35, 50]
}

def average(scores):
    sum_scores = 0
    for score in scores:
        sum_scores += score
    return round(sum_scores / len(scores), 1)

def letter_grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 80:
        return 'B'
    elif avg >= 70:
        return 'C'
    elif avg >= 60:
        return 'D'
    else:
        return 'F'

for student, scores in students.items():
    avg_score = average(scores)
    grade = letter_grade(avg_score)
    print(f'{student:<7} | Avg: {avg_score:<2} | Grade: {grade:<1}')
````
