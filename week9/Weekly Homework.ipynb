## Problem 1: Conditional sum calculator [10%]

Write a Python function `conditional_sum(*numbers)` that accepts an **arbitrary number of arguments** (using `*args`). The function should return:
- The sum of their squares if an even number of arguments is given.
- The sum of their cubes if an odd number of arguments is given.

If no arguments are passed, the function should return `0`.

### Example usage:
```python
    print(conditional_sum(1, 2, 3))      # Expected output: 36  (1^3 + 2^3 + 3^3 = 1 + 8 + 27 = 36)
    print(conditional_sum(4, 5))         # Expected output: 41  (4^2 + 5^2 = 16 + 25 = 41)
    print(conditional_sum())             # Expected output: 0   (No arguments passed)
    print(conditional_sum(7, 8, 9, 10))  # Expected output: 462 (7^2 + 8^2 + 9^2 + 10^2 = 49 + 64 + 81 + 100 = 294)

````python
def conditional_sum(*numbers):
    length = len(numbers)
    sum = 0
    
    if length % 2 == 0:
        for num in numbers:
            sum += num**2
    else:        
        for num in numbers:
            sum += num**3
    return sum
````

Then run the following cell to check if the output is the same as expected

````python
print(conditional_sum(1, 2, 3))      # Expected output: 36  (1^3 + 2^3 + 3^3 = 1 + 8 + 27 = 36)
print(conditional_sum(4, 5))         # Expected output: 41  (4^2 + 5^2 = 16 + 25 = 41)
print(conditional_sum())             # Expected output: 0   (No arguments passed)
print(conditional_sum(7, 8, 9, 10))  # Expected output: 462 (7^2 + 8^2 + 9^2 + 10^2 = 49 + 64 + 81 + 100 = 294)
````

## Problem 2: Student grade tracker [10%]

Write a Python function `calculate_average_grade()` that accepts an **arbitrary number of keyword arguments**, where each key-value pair represents name and grade. 

The function should return a tuple containing:
- The average grade of all students.
- The name of the top student (the one with the highest grade).

If nothing is passed, the function should return `(0, None)`.

### Example usage:
```python
    print(calculate_average_grade(Alice=90, Bob=85, Charlie=92))  # Expected output: (89.0, 'Charlie')
    print(calculate_average_grade(Mike=78, John=88))              # Expected output: (83.0, 'John')
    print(calculate_average_grade())                              # Expected output: (0, None)
    print(calculate_average_grade(Diana=100, Eva=95, Frank=80))   # Expected output: (91.6667, 'Diana')

````python
def calculate_average_grade(**grades):
    if len(grades) == 0:
        return (0, None)
    
    average = sum(grades.values()) / len(grades)
    top_student = max(grades, key=grades.get)

    return (average, top_student)
````

Then run the following cell to check if the output is the same as expected

````python
print(calculate_average_grade(Alice=90, Bob=85, Charlie=92))  # Expected output: (89.0, 'Charlie')
print(calculate_average_grade(Mike=78, John=88))              # Expected output: (83.0, 'John')
print(calculate_average_grade())                              # Expected output: (0, None)
print(calculate_average_grade(Diana=100, Eva=95, Frank=80))   # Expected output: (91.6667, 'Diana')
````

## Problem 4: Record scores [10%]

Write a Python function `record_scores()` that has two parameters
- **name**: a student's name as a positional argument 
- **scores**: an arbitrary number of scores 

The function should return a dictionary containing:
- The student's name.
- The average score of the provided scores.
- The highest score among the provided scores.

If no scores are provided, the function should return a dictionary with the average score set to `0` and the highest score set to `None`.

### Example usage:
```python
    print(record_scores("Alice", 90, 85, 88))       # Expected output: {'name': 'Alice', 'average_score': 87.67, 'highest_score': 90}
    print(record_scores("Bob", 78, 82))             # Expected output: {'name': 'Bob', 'average_score': 80.0, 'highest_score': 82}
    print(record_scores("Charlie"))                 # Expected output: {'name': 'Charlie', 'average_score': 0, 'highest_score': None}
    print(record_scores("Diana", 100, 95, 85, 90))  # Expected output: {'name': 'Diana', 'average_score': 92.5, 'highest_score': 100}

````python
def record_scores(name, *scores):
    if len(scores) == 0:
        average_score = 0
        highest_score = 0
    else:
        average_score = sum(scores) / len(scores)
        highest_score = max(scores)
    
    return {'name': name, 'average_score': average_score, 'highest_score': highest_score}
````

Then run the following cell to check if the output is the same as expected

````python
print(record_scores("Alice", 90, 85, 88))       # Expected output: {'name': 'Alice', 'average_score': 87.67, 'highest_score': 90}
print(record_scores("Bob", 78, 82))             # Expected output: {'name': 'Bob', 'average_score': 80.0, 'highest_score': 82}
print(record_scores("Charlie"))                 # Expected output: {'name': 'Charlie', 'average_score': 0, 'highest_score': None}
print(record_scores("Diana", 100, 95, 85, 90))  # Expected output: {'name': 'Diana', 'average_score': 92.5, 'highest_score': 100}
````

## Problem 5: Student performance report [10%]

Write a Python function `generate_report()` that has three parameters:
- **name**: a positional argument for the student's name.
- **scores**: an arbitrary number of positional arguments for the scores.
- **additional_info**: an arbitrary number of keyword arguments for additional information (e.g., `age`, `grade`, `attendance`).

The function should return a dictionary containing:
- The student's name.
- The average score of the provided scores.
- The highest score among the provided scores.
- Any additional information provided in the keyword arguments.

If no scores are provided, the function should set the average score to `0` and the highest score to `None`.
### Example usage:
```python
    print(generate_report("Alice", 90, 85, 88, age=16, grade="10th"))   # Expected output: {'name': 'Alice', 'average_score': 87.67, 'highest_score': 90, 'age': 16, 'grade': '10th'}
    print(generate_report("Bob", 78, 82, attendance="95%"))             # Expected output: {'name': 'Bob', 'average_score': 80.0, 'highest_score': 82, 'attendance': '95%'}
    print(generate_report("Charlie"))                                   # Expected output: {'name': 'Charlie', 'average_score': 0, 'highest_score': None}
    print(generate_report("Diana", 100, 95, 85, 90, grade="A", participation=True)) # Expected output: {'name': 'Diana', 'average_score': 92.5, 'highest_score': 100, 'grade': 'A', 'participation': True}

````python
def generate_report(name, *scores, **additional_info):
    if len(scores) == 0:
        average_score = 0
        highest_score = 0
    else:
        average_score = sum(scores) / len(scores)
        highest_score = max(scores)
    
    report = {
        'name': name,
        'average_score': average_score,
        'highest_score': highest_score
    }
    
    report.update(additional_info)
    
    return report
````

Then run the following cell to check if the output is the same as expected

````python
print(generate_report("Alice", 90, 85, 88, age=16, grade="10th"))   # Expected output: {'name': 'Alice', 'average_score': 87.67, 'highest_score': 90, 'age': 16, 'grade': '10th'}
print(generate_report("Bob", 78, 82, attendance="95%"))             # Expected output: {'name': 'Bob', 'average_score': 80.0, 'highest_score': 82, 'attendance': '95%'}
print(generate_report("Charlie"))                                   # Expected output: {'name': 'Charlie', 'average_score': 0, 'highest_score': None}
print(generate_report("Diana", 100, 95, 85, 90, grade="A", participation=True)) # Expected output: {'name': 'Diana', 'average_score': 92.5, 'highest_score': 100, 'grade': 'A', 'participation': True}
````

Assuming you already implemented `fact()` function in previous problem, now you need to compare its performance against `math` module's `factorial()` function for different values of `n`.

Your task is to create a `benchmark()` function
- it has two parameters
   - `func`: it's a function to be used. It can be either `math.factorial` or `math_utils.fact` 
   - `n`, a positive non-zero integer number
- it measures the execution time of `func` for given `n` using `perf_counter()` from `time` module.

### Example usage:

```python
   for n in [123, 1234, 12345]:
      print("Benchmarking for n =", n)
      print("  math.factorial took {} seconds".format(benchmark(math.factorial, n)))
      print("  math_utils.fact took {} seconds".format(benchmark(math_utils.fact, n)))
```
### Example output:
```terminal
   Benchmarking for n = 123:   
      math.factorial took 0.0012 seconds 
      math_utils.fact took 0.002 seconds 
   Benchmarking for n = 1234:   
      math.factorial took 0.012 seconds 
      math_utils.fact took 0.02 seconds 
   Benchmarking for n = 12345:   
      math.factorial took 0.12 seconds 
      math_utils.fact took 0.2 seconds 

define `benchmark()` function in the following cell

````python
import time

def benchmark(func, *args, **kwargs):
    start = time.time()
    func(*args, **kwargs)
    end = time.time()
    execution_time = end - start
    return execution_time
````
