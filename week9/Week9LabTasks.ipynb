## Task 2.3: Copying and moving files with shutil
### Description:

Create a directory *shutil_test*.
Inside it, create a file "original_file.txt" with any text content.
Copy "original_file.txt" to "copy_file.txt" using shutil.copy().
Move (rename) "copy_file.txt" to "copy_file_renamed.txt".
Print the contents of the directory to confirm your results.

### Hints:

shutil.copy(source, destination) copies a file.  
shutil.move(source, destination) moves or renames a file.

```python
# ==== Your solution for Task 2.3 here ====
import shutil

# 1. Create "shutil_test" directory
os.mkdir("shutil_test")
print("Created shutil_test directory")

# 2. Create "original_file.txt" with some text content
path = os.path.abspath("shutil_test") + '/original_file.txt'

with open(path, 'w') as f:
    f.write("Some content")

# 3. Copy it to "copy_file.txt"
path_copy = os.path.abspath("shutil_test") + '/copy_file.txt'
shutil.copy(path, path_copy)
print('Copied file to copy_file.txt')

# 4. Move "copy_file.txt" to "copy_file_renamed.txt"
path_copy_renamed = os.path.abspath("shutil_test") + '/copy_file_renamed.txt'
shutil.move(path_copy, path_copy_renamed)
print('Moved file to copy_file_renamed.txt')

# 5. Print contents to confirm
content = open(path_copy_renamed, 'r').read()
print(content)
```

### Task 2.4: Random data generation with random
## Description:

Generate 10 random integers between 1 and 100 (inclusive). Print them out.
Create a list of 5 random floating-point numbers between 0.0 and 1.0.
Randomly shuffle a list of your favorite fruits.
Print a random choice from a list of integers [10, 20, 30, 40, 50].

### Hints:

*random.randint(a, b)* to get a random integer.  
*random.random()* for floats between 0.0 and 1.0.  
*random.shuffle(list)* shuffles a list in place.  
*random.choice(sequence)* picks a random element.

```python
# ==== Your solution for Task 2.4 here ====
import random

# 1. Generate 10 random integers (1 to 100)
random_integers = []
for i in range(10):
    random_integers.append(random.randint(1, 100))

print("Random integers: ", random_integers)

# 2. Create a list of 5 random floats (0.0 to 1.0)
random_floats = []
for i in range(5):
    random_floats.append(random.random())

print("Random floats: ", random_floats)

# 3. Shuffle a list of fruits
fruits = ['apple', 'banana', 'stawberry', 'orange']
random.shuffle(fruits)
print(fruits)

# 4. Print a random choice from [10, 20, 30, 40, 50]
nums = [10, 20, 30, 40, 50]
print(random.choice(nums))
```
